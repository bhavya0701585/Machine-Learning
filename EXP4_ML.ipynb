```python

import numpy as np

np.random.seed(42)

# XOR dataset
X = np.array([[0, 0],
              [0, 1],
              [1, 0],
              [1, 1]])

Y = np.array([[0], [1], [1], [0]])

# Initialize weights and biases
W1 = np.random.randn(2, 4)
b1 = np.zeros((1, 4))
W2 = np.random.randn(4, 1)
b2 = np.zeros((1, 1))

learning_rate = 0.5
epochs = 10000

def sigmoid(x):
    return 1 / (1 + np.exp(-np.clip(x, -500, 500)))

# Train using backpropagation
for epoch in range(epochs):
    # Forward propagation
    H = sigmoid(np.dot(X, W1) + b1)
    output = sigmoid(np.dot(H, W2) + b2)

    # Calculate error
    error = Y - output

    # Backpropagation
    d_output = error * output * (1 - output)
    d_hidden = np.dot(d_output, W2.T) * H * (1 - H)

    # Update weights and biases
    W2 += learning_rate * np.dot(H.T, d_output) / len(X)
    b2 += learning_rate * np.mean(d_output, axis=0, keepdims=True)
    W1 += learning_rate * np.dot(X.T, d_hidden) / len(X)
    b1 += learning_rate * np.mean(d_hidden, axis=0, keepdims=True)

# Test the trained network
H = sigmoid(np.dot(X, W1) + b1)
output = sigmoid(np.dot(H, W2) + b2)

print("XOR Predictions:")
for i in range(len(X)):
    print(X[i], "->", round(output[i, 0], 4),
          "Class:", int(output[i, 0] >= 0.5))

print("\nFinal Mean Squared Error:",
      np.mean((Y - output) ** 2))
```
