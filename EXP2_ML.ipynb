```python

import pandas as pd
from itertools import product

# Create and save training data
data = [
    ["Sunny", "Warm", "Normal", "Strong", "Warm", "Same", "Yes"],
    ["Sunny", "Warm", "High", "Strong", "Warm", "Same", "Yes"],
    ["Rainy", "Cold", "High", "Strong", "Warm", "Change", "No"],
    ["Sunny", "Warm", "High", "Strong", "Cool", "Change", "Yes"]
]

columns = ["Sky", "Temperature", "Humidity", "Wind",
           "Water", "Forecast", "EnjoySport"]

df = pd.DataFrame(data, columns=columns)
df.to_csv("training_data.csv", index=False)

# Read CSV
df = pd.read_csv("training_data.csv")
X = df.iloc[:, :-1].values.tolist()
y = df.iloc[:, -1].tolist()

# Generate all hypotheses
domains = [df[col].unique().tolist() for col in columns[:-1]]
all_hypotheses = list(product(*[d + ["?"] for d in domains]))

def matches(h, x):
    return all(a == "?" or a == b for a, b in zip(h, x))

# Keep hypotheses consistent with every example
consistent = [
    h for h in all_hypotheses
    if all(matches(h, x) == (label == "Yes")
           for x, label in zip(X, y))
]

def more_general(h1, h2):
    return all(a == "?" or a == b for a, b in zip(h1, h2))

# Find maximally specific and maximally general boundaries
S = [
    h for h in consistent
    if not any(h != other and more_general(h, other)
               for other in consistent)
]

G = [
    h for h in consistent
    if not any(h != other and more_general(other, h)
               for other in consistent)
]

print("Training Data:")
print(df.to_string(index=False))

print("\nSpecific Boundary (S):")
for h in S:
    print(h)

print("\nGeneral Boundary (G):")
for h in G:
    print(h)

print("\nNumber of consistent hypotheses:", len(consistent))
print("All consistent hypotheses:")
for h in consistent:
    print(h)
```
